```python
import matplotlib.pyplot as plt
```

```python
x=[2,4,6,8,10]
y=[1,3,4,5,6]
y1=[5,6,7,8,9]
x1=[5,6,7,8,10]
plt.plot(x,y)
plt.show()
```

```python
#individual graphs
plt.subplot(2,2,1)
plt.plot(x,y)

plt.subplot(2,2,2)
plt.plot(x,y1)

plt.show()
```

```python
#both lines in one graph
plt.subplot(2,2,1)
plt.plot(x,y)

plt.subplot(2,2,1)
plt.plot(x,y1)

plt.show()
```

```python

plt.subplot(2,2,1)
plt.title("Graph-1")      #title for graph
plt.suptitle("Data")      # main/super title for all graphs
plt.xlabel("x-axis")      #label for x axis
plt.ylabel("y-axis")      #label for y axis
plt.plot(x,y)

plt.subplot(2,2,2)
plt.title("Graph-2")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.plot(x1,y1)

plt.subplot(2,2,3)
plt.title("Graph-3")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.plot(x1,y)

plt.subplot(2,2,4)
plt.title("Graph-4")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.plot(x,y1)

plt.show()
```
